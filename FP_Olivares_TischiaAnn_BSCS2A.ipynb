## Performance Analysis of Optimizers for Fashion MNIST Classification Using CNN

### **1. Introduction**
- This will analyze the performance of differrent Deep Learning Optimization Methods: SDG, Adam, and RMSProp on a CNN model for Fashion-MNIST.

### **2. Data Preparation**
- Loading the Fashion-MNIST dataset from Keras.
- Split the data into training and testing
- Normalizing and Reshaping data

```python
# Prepare necessary Libraries
from keras.datasets import fashion_mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Dense, Flatten
import matplotlib.pyplot as plt
import numpy as np
from tensorflow.keras.layers import Input

# Split the data into training and testing
(trainX, trainy), (testX, testy) = fashion_mnist.load_data()

# Visualizing some of the sample images from the dataset
for i in range(1, 10):
    plt.subplot(3, 3, i)
    plt.imshow(trainX[i], cmap=plt.get_cmap('gray'))
plt.show()

# Normalize the input data
trainX = trainX / 255.0
testX = testX / 255.0

# Reshaping data to include the channel dimension
trainX = np.expand_dims(trainX, -1)
testX = np.expand_dims(testX, -1) 
```

### **3. CNN Architecture Model**
-  a straightforward CNN architecture with three convolutional layers to perform convolutional operation and extract features, and three max-pooling layer to downsample the features.

```python
def model_arch():
    models = Sequential()
     
    # Input layer
    models.add(Input(shape=(28, 28, 1)))
    
    # Convolutional and pooling operations
    models.add(Conv2D(64, (5, 5),
                      padding="same",
                      activation="relu"))
     
    models.add(MaxPooling2D(pool_size=(2, 2)))
    models.add(Conv2D(128, (5, 5), padding="same",
                      activation="relu"))
     
    models.add(MaxPooling2D(pool_size=(2, 2)))
    models.add(Conv2D(256, (5, 5), padding="same", 
                      activation="relu"))
     
    models.add(MaxPooling2D(pool_size=(2, 2)))
     
    # Flatten layer and fully connected layers are added
    models.add(Flatten())
    models.add(Dense(256, activation="relu"))
    
    # 10 is created with a softmax activation function
    models.add(Dense(10, activation="softmax"))
    return models
```

### **4. Training Process with Different Optimizers**

#### 4.1 Stochastic Gradient Descent (SGD) with Momentum

```python
from tensorflow.keras.optimizers import SGD
modelSGD = model_arch()
 
modelSGD.compile(optimizer=SGD(learning_rate=0.001, momentum=0.9),
              loss='sparse_categorical_crossentropy',
              metrics=['sparse_categorical_accuracy'])

historySGD = modelSGD.fit(
    trainX.astype(np.float32), trainy.astype(np.float32),
    epochs=10,
    validation_split=0.33,
    batch_size=32
)
```

#### 4.2 Adam (Adaptive Moment Estimation)

```python
from tensorflow.keras.optimizers import Adam
modelAdam = model_arch()
 
modelAdam.compile(optimizer=Adam(learning_rate=0.001),
              loss='sparse_categorical_crossentropy',
              metrics=['sparse_categorical_accuracy'])

historyAdam = modelAdam.fit(
    trainX.astype(np.float32), trainy.astype(np.float32),
    epochs=10,
    validation_split=0.33,
    batch_size=32
)
```

#### 4.3 RMSProp (Root Mean Square Propagation)

```python
from tensorflow.keras.optimizers import RMSprop
modelRMSprop = model_arch()
 
modelRMSprop.compile(optimizer=RMSprop(learning_rate=0.001 ),
              loss='sparse_categorical_crossentropy',
              metrics=['sparse_categorical_accuracy'])

historyRMSprop = modelRMSprop.fit(
    trainX.astype(np.float32), trainy.astype(np.float32),
    epochs=10,
    validation_split=0.33,
    batch_size=32
)
```

### **5. Testing the Different Optimizers**

5.1 SGD Testing

```python
#Test Accuracy of SGD
scoreSGD = modelSGD.evaluate(testX, testy, verbose=1)
print(scoreSGD)
```

5.2 Adam Testing

```python
#Test Accuracy of Adam
scoreAdam = modelAdam.evaluate(testX, testy, verbose=1)
print(scoreAdam)
```

5.3 RMSProp Testing

```python
#Test Accuracy of RMSProp
scoreRMSProp = modelRMSprop.evaluate(testX, testy, verbose=1)
print(scoreRMSProp)
```

### **6. Performance Analysis**

6.1 SGD Training Accuracy and Loss

```python
plt.figure(figsize=(12, 5)) 

# Accuracy vs Epoch plot
plt.subplot(1, 2, 1)
plt.plot(historySGD.history['sparse_categorical_accuracy'], label='Train Accuracy')
plt.plot(historySGD.history['val_sparse_categorical_accuracy'], label='Validation Accuracy')
plt.title('SGD Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epochs')
plt.legend()

# Loss vs Epoch plot
plt.subplot(1, 2, 2)
plt.plot(historySGD.history['loss'], label='Train Loss')
plt.plot(historySGD.history['val_loss'], label='Validation Loss')
plt.title('SGD Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epochs')
plt.legend()

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 5)) 

# Accuracy vs Epoch plot
plt.subplot(1, 2, 1)
plt.plot(historyAdam.history['sparse_categorical_accuracy'], label='Train Accuracy')
plt.plot(historyAdam.history['val_sparse_categorical_accuracy'],label='Validation Accuracy')
plt.title('Adam Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('epoch')
plt.legend()

# Loss vs Epoch plot
plt.subplot(1, 2, 2)
plt.plot(historyAdam.history['loss'], label='Train Loss')
plt.plot(historyAdam.history['val_loss'], label='Validation Loss')
plt.title('Adam Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epochs')
plt.legend()

plt.tight_layout()
plt.show()
```

6.3 RMSprop

```python
plt.figure(figsize=(12, 5)) 

# Accuracy vs Epoch plot
plt.subplot(1, 2, 1)
plt.plot(historyRMSprop.history['sparse_categorical_accuracy'], label='Train Accuracy')
plt.plot(historyRMSprop.history['val_sparse_categorical_accuracy'],label='Validation Accuracy')
plt.title('RMSprop Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epochs')
plt.legend()

# Loss vs Epoch plot
plt.subplot(1, 2, 2)
plt.plot(historyRMSprop.history['loss'], label='Train Loss')
plt.plot(historyRMSprop.history['val_loss'], label='Validation Loss')
plt.title('RMSprop Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epochs')
plt.legend()

plt.tight_layout()
plt.show()
```

### **7. Predictions from Each Optimizer**

10 Random Test Images

```python
import random

# Select 10 random indices from the test dataset
random_indices = random.sample(range(len(testX)), 10)

plt.figure(figsize=(12, 6))

# Visualize the images
for i in range(10):  
    plt.subplot(2, 5, i + 1) 
    plt.imshow(testX[random_indices[i]], cmap=plt.get_cmap('gray'))

plt.tight_layout()
plt.show()

labels = ['t_shirt', 'trouser', 'pullover', 'dress', 'coat',
          'sandal', 'shirt', 'sneaker', 'bag', 'ankle_boots']
```

7.1 SGD Optimizer with SGD Prediction

```python
# Make a prediction using the CNN Model with SDG Optimizer
predictionsSGD = modelSGD.predict(testX[random_indices])
predicted_labels_SGD = [labels[np.argmax(pred)] for pred in predictionsSGD]
 
print("Predicted Labels Using SDG Optimizer:")
for i, label in enumerate(predicted_labels_SGD):
    print(f"Image {i + 1}: {label}")
```

7.2 Adam Optimizer with CNN Prediction

```python
# Make a prediction using the CNN Model with SDG Optimizer
predictionsAdam = modelAdam.predict(testX[random_indices])
predicted_labels_Adam = [labels[np.argmax(pred)] for pred in predictionsAdam]
 
print("Predicted Labels Using Adam Optimizer:")
for i, label in enumerate(predicted_labels_Adam):
    print(f"Image {i + 1}: {label}")
```

7.3 RMSProp Optimizer with CNN Prediction

```python
# Make a prediction using the CNN Model with SDG Optimizer
predictionsRMSprop = modelRMSprop.predict(testX[random_indices])
predicted_labels_RMSprop = [labels[np.argmax(pred)] for pred in predictionsRMSprop]
 
print("Predicted Labels Using RMSProp Optimizer:")
for i, label in enumerate(predicted_labels_RMSprop):
    print(f"Image {i + 1}: {label}")
```

### **8. Compare Predictions with True Labels**

```python
# Get the true labels for the selected test images
true_labels = testy[random_indices]

# Compare predictions three optimizers with true labels
print("Prediction Comparison:")
for i, (true, pred_SGD, pred_Adam, pred_RMSProp) in enumerate(zip(true_labels, predicted_labels_SGD, predicted_labels_Adam, predicted_labels_RMSprop)):
    true_label = labels[true]
    print(f"Image {i + 1}: True: {true_label}")
    print(f"  Predicted using SGD: {pred_SGD} ({'Correct' if pred_SGD == true_label else 'Incorrect'})")
    print(f"  Predicted using Adam: {pred_Adam} ({'Correct' if pred_Adam == true_label else 'Incorrect'})")
    print(f"  Predicted using RMSProp: {pred_RMSProp} ({'Correct' if pred_RMSProp == true_label else 'Incorrect'})")
```
